```python
funnel = {
    "Applied": df["application_date"].notna().sum(),
    "Screened": df["screening_date"].notna().sum(),
    "Shortlisted": df["shortlist_date"].notna().sum(),
    "Interviewed": df["interview_date"].notna().sum(),
    "Selected": df["selection_date"].notna().sum(),
    "Offered": df["offer_date"].notna().sum(),
    "Joined": df["joining_date"].notna().sum()
}

funnel
```

```python
funnel_df = pd.DataFrame(
    list(funnel.items()),
    columns=["Stage", "Candidates"]
)

funnel_df
```

```python
funnel_df["Conversion_Rate"] = (
    funnel_df["Candidates"] /
    funnel_df["Candidates"].shift(1)
) * 100
```

```python
funnel_df
```

```python
overall_hiring_rate = (
    funnel["Joined"] / funnel["Applied"]
) * 100

overall_hiring_rate
```

```python
total_applicants = df["application_date"].notna().sum()
total_joined = df["joining_date"].notna().sum()

print("Total Applicants:", total_applicants)
print("Total Joined:", total_joined)
```

```python
avg_time_to_hire = df["time_to_hire"].mean()

print("Average Time to Hire:", round(avg_time_to_hire, 2), "days")
```

```python
avg_time_to_fill = df["time_to_fill"].mean()

print("Average Time to Fill:", round(avg_time_to_fill, 2), "days")
```

```python
avg_interview_score = df["overall_interview_score"].mean()

print(
    "Average Interview Score:",
    round(avg_interview_score, 2)
)
```

```python
df["offer_status"].value_counts()
```

```python
accepted_offers = (df["offer_status"] == "Accepted").sum()
rejected_offers = (df["offer_status"] == "Rejected").sum()

offer_acceptance_rate = (
    accepted_offers /
    (accepted_offers + rejected_offers)
) * 100

print("Offer Acceptance Rate:", round(offer_acceptance_rate, 2), "%")
```

```python
total_selected = df["selection_date"].notna().sum()

hiring_success_rate = (
    total_joined / total_selected
) * 100

print("Total Selected:", total_selected)
print("Hiring Success Rate:", round(hiring_success_rate, 2), "%")
```

```python
kpi_df = pd.DataFrame({
    "KPI": [
        "Total Applicants",
        "Total Joined",
        "Average Time to Hire",
        "Average Time to Fill",
        "Average Interview Score",
        "Offer Acceptance Rate",
        "Hiring Success Rate"
    ],
    "Value": [
        total_applicants,
        total_joined,
        round(avg_time_to_hire, 2),
        round(avg_time_to_fill, 2),
        round(avg_interview_score, 2),
        round(offer_acceptance_rate, 2),
        round(hiring_success_rate, 2)
    ]
})

kpi_df
```

```python
kpi_df.to_csv(
    "../datasets/recruitment_kpis.csv",
    index=False
)
```

```python
best_source = source_analytics["Hiring_Rate"].idxmax()
best_rate = source_analytics["Hiring_Rate"].max()

print("Best Recruitment Source:", best_source)
print("Hiring Rate:", best_rate, "%")
```

```python
source_score = (
    df.groupby("source")["overall_interview_score"]
    .mean()
    .round(2)
    .sort_values(ascending=False)
)

source_score
```

```python
df["match_category"] = pd.cut(
    df["match_score"],
    bins=[0, 59, 74, 89, 100],
    labels=["Low", "Average", "Good", "Excellent"]
)
```

```python
match_distribution = (
    df["match_category"]
    .value_counts()
    .sort_index()
)

match_distribution
```

```python
match_effectiveness = (
    df.groupby("match_category", observed=True)
    .agg(
        Applicants=("candidate_id", "count"),
        Joined=("joining_date", lambda x: x.notna().sum())
    )
)

match_effectiveness["Hiring_Rate"] = (
    match_effectiveness["Joined"] /
    match_effectiveness["Applicants"]
) * 100

match_effectiveness["Hiring_Rate"] = (
    match_effectiveness["Hiring_Rate"].round(2)
)

match_effectiveness
```

```python
match_effectiveness.to_csv(
    "../datasets/match_effectiveness.csv"
)
```

```python
df["experience_category"] = pd.cut(
    df["experience_years"],
    bins=[-1, 1, 3, 5, 10, float("inf")],
    labels=[
        "0-1 Years",
        "2-3 Years",
        "4-5 Years",
        "6-10 Years",
        "10+ Years"
    ]
)
```

```python
experience_effectiveness = (
    df.groupby("experience_category", observed=True)
    .agg(
        Applicants=("candidate_id", "count"),
        Joined=("joining_date", lambda x: x.notna().sum())
    )
)

experience_effectiveness["Hiring_Rate"] = (
    experience_effectiveness["Joined"] /
    experience_effectiveness["Applicants"]
) * 100

experience_effectiveness["Hiring_Rate"] = (
    experience_effectiveness["Hiring_Rate"].round(2)
)

experience_effectiveness
```

```python
experience_effectiveness.to_csv(
    "../datasets/experience_effectiveness.csv"
)
```
